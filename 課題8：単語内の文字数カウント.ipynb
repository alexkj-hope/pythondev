```python
words = []

letters = "abcdefghijklmnopqrstuvwxyz"
counts = {ch: 0 for ch in letters}

while True:
    w = input("英単語を入力してください：").strip()
    if w == "":
        break
    words.append(w)

    for ch in w.lower():
        if 'a' <= ch <= 'z':
            counts[ch] += 1

sorted_words = sorted(words, key=str.lower)
print(f"入力した英単語：{sorted_words}")

for ch in letters:
    c = counts[ch]
    if c >= 1:
        print(f"{ch}が{c}個ありました")
```
